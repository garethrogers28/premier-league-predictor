#### Load all datasets into memory

Now that we have confirmed the structure of a single file, we loop through all CSV files in `data/raw` and load each one into a list of DataFrames. Each element in `season_data` represents one Premier League season.

```python
season_data = []
```

```python
for file_name in csv_files:
    file_path = os.path.join(data_path, file_name)
    df = pd.read_csv(file_path)
    season_data.append(df)
```

```python
len(season_data)
```

```python
type(season_data[0])
```

```python
season_data[0].shape
```

#### Check for column inconsistencies across seasons

The datasets are compared to identify any structural differences before merging. This includes checking for missing or additional columns, differences in column names, and inconsistencies in column order across seasons.

```python
for i, season_df in enumerate(season_data):
    print(i, season_df.shape)
```

```python
set(season_data[0].columns) - set(season_data[6].columns)
```

#### Remove unnecessary index column

Some CSV files were exported with the DataFrame index saved as a column called `Unnamed: 0`. This column is not a player statistic and is removed before merging the datasets.

```python
set(season_data[6].columns) - set(season_data[0].columns)
```

#### Standardise column names

Although the datasets contained the same features, three column names used different capitalisation between seasons (e.g. `Big chances created` and `Big Chances Created`). These column names were standardised to ensure a consistent schema before merging the datasets.

```python
for i, season_df in enumerate(season_data):
    print(i, list(season_df.columns))
```

```python
for season_df in season_data:
    if 'Unnamed: 0' in season_df.columns:
        season_df.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
for i, season_df in enumerate(season_data):
    print(i, season_df.shape)
```

```python
for season_df in season_data:
    season_df.rename(
        columns={
            "Big Chances Created": "Big chances created",
            "Goals Conceded": "Goals conceded",
            "Penalties Saved": "Penalties saved",
        },
        inplace=True,
    )
```

```python
season_data[7].columns
```

```python
season_data[0].columns
```

### Align column order across all seasons

Although all datasets contain the same columns, one season stores them in a different order. To ensure a consistent dataset structure before merging, the column order is aligned using the first season as the reference.

```python
season_data[0].columns[3]
```

```python
season_data[7].columns[3]
```

```python
for i, (col1, col2) in enumerate(zip(season_data[0].columns, season_data[7].columns)):
    if col1 != col2:
        print(f"{i}: {col1} -> {col2}")
```

```python
season_data[7] = season_data[7][season_data[0].columns]
season_data[7].columns.equals(season_data[0].columns)
```

```python
reference_columns = season_data[0].columns

for index, season_df in enumerate(season_data):
    season_data[index] = season_df[reference_columns]
```

```python
for index, season_df in enumerate(season_data):
    print(index, season_df.columns.equals(reference_columns))
```
